# **Now the Gradient based Descent**

```python
w=0
b=0
plt.scatter(x,y)
for i in range(6):

    yp = w * x + b
    plt.plot(x, yp)
    mse = np.mean((y - yp)**2)

    deriv_dw = (-2) * np.mean(x * (y - yp))
    deriv_db = (-2) * np.mean(y - yp)

    # Update weights and intercept
    # w = w + 0.001 * (mse)
    # b = b + 0.01 * (mse)
    b = b - 0.01*deriv_db
    w = w - 0.01*deriv_dw

    yp = w*x + b
    plt.plot(x,yp)

    print(mse,w,b)

```
